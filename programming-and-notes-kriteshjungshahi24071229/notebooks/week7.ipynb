# Concepts Covered

Introduction to Sets
A set is a collection of values, similar to a list or tuple, but with some important differences.
A value can appear only once (no duplicates)
Sets are unordered
Sets contain immutable values
Sets do not support indexing or slicing
Sets support membership testing and set operations
Sets are iterable and can be used in loops

# Creating A Set

Sets are written using curly braces {}.

The order may change because sets are unordered.

If a value appears more than once, it will only exist once in the set.

```python
vowels = {"a", "e", "i", "o", "u"}
print(vowels)
```

```python
numbers = {1, 2, 2, 3, 3}
print(numbers)
```

# Creating An Empty Set

An empty set cannot be created using {} because that creates a dictionary.

```python
empty_set = set()
```

# Creating A Set From An Iterable

The set() constructor can be used to convert another iterable into a set.

Only one value is passed to the constructor.

```python
letters = set("aeiou")
print(letters)
```

# Working With Sets

Sets support common mathematical operations.

```python
set1 = {1, 2, 3}
set2 = {3, 4, 5}

print(set1 | set2)   # Union
print(set1 & set2)   # Intersection
print(set1 - set2)   # Difference
```

# Introduction To Dictionaries

A dictionary stores data as key–value pairs.

Keys must be unique

Values can be of any data type

Dictionaries are unordered (conceptually)

```python
student = {
    "name": "Alex",
    "age": 20,
    "course": "Computing"
}
```

# Acessing And Modifying Dicitionaries

Values are accessed using keys.

Values can be updated or added.

```python
print(student["name"])
```

```python
student["age"] = 21
student["year"] = 1
```

# Dictionary Methods

Some common dictionary methods include:

```python
print(student.keys())
print(student.values())
print(student.items())
```
