# Concepts Covered

Boolean Expressions
Boolean expressions evaluate to either True or False
They are used to control decisions in a program

# Example:
price > 0
len(password) >= 9
answer == correct_character

# Conditional Statements

Conditional statements allow a program to choose different actions based on conditions.
Common conditional statements:

if

if-else

if-elif-else

# Example

```python
if age >= 18:
    print("Allowed")
else:
    print("Not allowed")
```

# Iteration (Loops)

Loops allow instructions to be repeated until a condition is met.
Types of loops used:

while loop → repeat until condition becomes false

for loop → repeat for a fixed number of times
