# Function is a reusable block of code that performs a specific task.
Python already provides many built-in functions such as print(), input(), range(), and type().

Example 1 : Using Externally Defined Functions

Functions that are not built into Python must be imported before they can be used.
These functions are usually stored inside modules.
It is good practice to only import what we need, to keep the program clean and easy to understand.

```python
import math
print(math.sqrt(25))
```

 Example 2 : Defining Our Own Functions,
 
We can define our own functions using the def keyword.

```python
def greet(name):
    print("Hello", name)

greet("Alex")
```

Example 3 : Default and Keyword Arguments

Functions can have default values, which are used if no value is passed.

```python
def greet(name="User"):
    print("Hello", name)

greet()
greet("Sam")
```

Example 4 : Argument Lists

Functions can accept multiple values using argument lists.

```python
def add_numbers(*numbers):
    print(sum(numbers))

add_numbers(2, 4, 6)
```

Example 5 : Lambda Expressions

Lambda expressions are small anonymous functions written in one line.

```python
square = lambda x: x * x
print(square(5))
```
