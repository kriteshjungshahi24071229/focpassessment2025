# Formatting Output

Python provides several ways to format output to make it more readable.

# Formatted String Literals (f-strings)

F-strings allow values to be embedded directly into strings.

This is the most modern and readable formatting method.

```python
name = "Alex"
age = 20
print(f"My name is {name} and I am {age} years old.")
```

# str.format() Method

This method inserts values into placeholders.

```python
print("My name is {} and I am {} years old.".format(name, age))
```

# %-Style Formatting

This is an older formatting method but still supported.

```python
print("My name is %s and I am %d years old." % (name, age))
```

# Introduction to File Handling

Files are used to store data permanently.

Common file types include:

Text files (.txt)

Data files (.csv)

Program files (.py)

# Writing to a File

To write to a file, we open it in write mode ("w").

```python
file = open("example.txt", "w")
file.write("Hello, this is a file.")
file.close()
```

# Reading from a File

To read from a file, we use read mode ("r").

```python
file = open("example.txt", "r")
content = file.read()
print(content)
file.close()
```

# Using with Statement

The with statement automatically closes the file.

```python
with open("example.txt", "r") as file:
    print(file.read())
```

# Controlling File Access Position

The seek() method changes the current file position.

```python
with open("example.txt", "r") as file:
    file.seek(0)
    print(file.read(5))
```
