# Concepts Covered
Lists Recap
Lists store multiple values in an ordered way.

# Example

```python
squares = [4, 9, 16, 25]
```

# Lists Supports:
Indexing
Slicing
Iteration using loops
Lists are mutable, meaning we can change them after creation.

# Method
A method is like a function, but it belongs to a specific data type.
Methods are called using:

```python
object.method()
```

# List Methods
Some common list methods include:
append()
extend()
insert()
remove()
pop()
clear()
sort()
reverse()

# Example

```python
numbers = [1, 2, 3]
numbers.extend([4, 5])
print(numbers)
```

# Lists Comprehension

Lists Comprehensions provide a shorter way to create lists.

# Example 

```python
squares = [x*x for x in range(1, 6)]
```

# Tuples 

Tuples are similar to lists but immutable they cannot be changed. They are useful when data should not be modified.

# Example

```python
coordinates = (10, 20)
```

# List Vs Tuples

Lists - mutable, more flexible

Tuples - immutable, safer for fixed data
