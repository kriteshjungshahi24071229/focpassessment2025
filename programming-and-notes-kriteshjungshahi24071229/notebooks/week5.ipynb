# Modules
A module is a file that contains Python code (functions, variables, classes).
Modules help us reuse code across different programs.

```python
import math
print(math.pi)
```

# Accesing Command Line Arguments

Python allows passing values when running a script using the command line.
These values are accessed using sys.argv.

```python
import sys

numbers = [float(x) for x in sys.argv[1:]]
average = sum(numbers) / len(numbers)
print("Average is", average)
```

Running The Script

```python
python average.py 10.2 8.8 2.6
```
